```python
import pandas as pd
```

```python
orders = pd.read_csv("/content/drive/MyDrive/Orders.csv")
products = pd.read_csv("/content/drive/MyDrive/Products.csv")
customers = pd.read_csv("/content/drive/MyDrive/Customers.csv")
```

```python
# In Pandas, concat is used to combine DataFrames along an axis.
```

```python
orders_part1 = orders.iloc[:60]
orders_part2 = orders.iloc[60:]
```

```python
print("Part 1:", orders_part1.shape)
print("Part 2:", orders_part2.shape)
```

```python
orders_combined = pd.concat(
    [orders_part1, orders_part2]
)
```

```python
# here we concatinated both parts of orders dataset
```

```python
display(orders_combined.head())
```

```python
# concatinating with the help of axis
# axis=0 -> its default , it means its concatenating rows
# axis=0 is vertical/row wise stacking
```

```python
orders_combined = pd.concat(
    [orders_part1, orders_part2],
    axis=0
)
```

```python
orders_combined.head()
```

```python
orders_combined.tail()
```

```python
# ignore_index=True -> Ignore the existing indexes of the input DataFrames and create a new continuous index.
```

```python
orders_part1=orders.iloc[:60].copy()
orders_part2=orders.iloc[60:].copy()

orders_combined = pd.concat(
    [orders_part1,orders_part2],
    axis=0,
    ignore_index=True
)
print("Original Orders shape: ", orders.shape)
print("part 1 shape:", orders_part1.shape)
print("part 2 shape:", orders_part2.shape)
print("Combined Orders shape:", orders_combined.shape)
```

```python
#axis=0 vs axis=1
```

```python
# axis=0 — rows
pd.concat([orders_part1, orders_part2],
          axis=0
          )
```

```python
# axis=1 — column-wise / horizontal concatenation
```

```python
# Select the first three columns from the Orders DataFrame
orders_columns_part1 = orders.iloc[:, :3]
```

```python
# Select the remaining columns from the Orders DataFrame
orders_columns_part2 = orders.iloc[:, 3:]
```

```python
#Concatenate them horizontally
```

```python
orders_com=pd.concat(
    [orders_columns_part1, orders_columns_part2],
    axis=1
)
```

```python
display(orders_com.head())
```

```python
# Compare the original and reconstructed DataFrame shapes
print("Original Orders shape:", orders.shape)
print("Reconstructed shape:", orders_com.shape)
```

```python
orders_customers=pd.merge(
    orders,
    customers,
    on="Customer_ID",
    how="inner"
)
# An inner join keeps only records where the key exists in both DataFrames.
```

```python
# Keep only matching Customer_ID values from both DataFrames
orders_customers_inner = pd.merge(
    orders,
    customers,
    on="Customer_ID",
    how="inner"
)
orders_customers_inner
```

```python
# Left Join
```

```python
# Keep all Orders and add matching Customer information where available
orders_customers_left = pd.merge(
    orders,
    customers,
    on="Customer_ID",
    how="left"
)
```

```python
orders_customers_left
```

```python
# Keep all Customers and add matching Order information where available
orders_customers_right = pd.merge(
    orders,
    customers,
    on="Customer_ID",
    how="right"
)
orders_customers_right
# keeps all customers
```

```python
# Outer Join
```

```python
# Keep all records from both Orders and Customers
orders_customers_outer = pd.merge(
    orders,
    customers,
    on="Customer_ID",
    how="outer"
)
orders_customers_outer
```

```python
# Merge Orders with Customers using Customer_ID
# A left join keeps every order and adds matching customer information
orders_customers = pd.merge(
    orders,
    customers,
    on="Customer_ID",
    how="left"
)
```

```python
# Display the first five rows of the merged DataFrame
display(orders_customers.head())
```

```python
# Check the shape after merging Orders and Customers
print("Orders shape:", orders.shape)
print("Orders + Customers shape:", orders_customers.shape)
```

```python
# Merge the Orders + Customers DataFrame with Products
# Product_ID is used as the common key
# A left join keeps every order
processed = pd.merge(
    orders_customers,
    products,
    on="Product_ID",
    how="left"
)
```

```python
# Display the first five rows of the final merged DataFrame
display(processed.head())
```

```python
# Display the shape after adding product information
print("Orders + Customers + Products shape:", processed.shape)
```

```python
# Display all columns in the final merged DataFrame
print(processed.columns.tolist())
```

```python
# Create the first half of the Orders DataFrame
orders_part1 = orders.iloc[:60].copy()

# Create the second half of the Orders DataFrame
orders_part2 = orders.iloc[60:].copy()

# Combine both parts vertically using concat()
# axis=0 means rows are combined vertically
# ignore_index=True creates a new continuous index
orders_combined = pd.concat(
    [orders_part1, orders_part2],
    axis=0,
    ignore_index=True
)
# Display the shapes to verify the concatenation
print("Original Orders shape:", orders.shape)
print("First part shape:", orders_part1.shape)
print("Second part shape:", orders_part2.shape)
print("Combined Orders shape:", orders_combined.shape)
```

```python
# Display the first five rows of the combined Orders DataFrame
display(orders_combined.head())
```

```python
# Merge Orders with Customers
# Merge Orders with Customers using Customer_ID
# A left join keeps all order records and adds matching customer information
orders_customers = pd.merge(
    orders_combined,
    customers,
    on="Customer_ID",
    how="left"
)
```

```python
# Display the first five rows after merging Orders and Customers
display(orders_customers.head())

# Display the shape of the merged DataFrame
print("Orders shape:", orders_combined.shape)
print("Orders + Customers shape:", orders_customers.shape)
```

```python
# Merge with Products
```

```python
# Merge the Orders + Customers DataFrame with the Products dataset
# Product_ID is used as the common key
# A left join keeps all order records
processed = pd.merge(
    orders_customers,
    products,
    on="Product_ID",
    how="left"
)
```

```python
# Display the first five rows of the fully merged DataFrame
display(processed.head())

# Display the shape of the final merged DataFrame
print("Final merged shape:", processed.shape)
```

```python
# Display all columns available after merging the three datasets
print(processed.columns.tolist())
```

```python
# Convert Order_Date to DateTime
```

```python
processed["Order_Date"]=pd.to_datetime(processed['Order_Date'])
```

```python
print(processed["Order_Date"].dtype)
```

```python
# Extract Year
```

```python
# Extract the year from Order_Date and create a new column
processed["Order_Year"]=processed["Order_Date"].dt.year
```

```python
# Extract Month
```

```python
# Extract the month number from Order_Date
processed["Order_Month"]=processed["Order_Date"].dt.month
```

```python
# Extract Month Name
```

```python
# Extract the month name from Order_Date
processed["Order_Month_Name"] = processed["Order_Date"].dt.month_name()
```

```python
# Extract Day
processed["Order_Day"] = processed["Order_Date"].dt.day
```

```python
# Extract the day name such as Monday, Tuesday, etc.
processed["Order_Day_Name"] = processed["Order_Date"].dt.day_name()
```

```python
# Display the original date and all newly extracted date information
display(
    processed[
        [
            "Order_Date",
            "Order_Year",
            "Order_Month",
            "Order_Month_Name",
            "Order_Day",
            "Order_Day_Name"
        ]
    ].head(10)
)
```

```python
# Use apply() to Calculate Total Amount
```

```python
# Total Amount = Unit Price × Quantity
```

```python
# Calculate the total value of each order
# The function is applied row-wise using axis=1
processed["Total_Amount"] = processed.apply(
    lambda row: row["Unit_Price"] * row["Quantity"],
    axis=1
)
```

```python
display(
    processed["Total_Amount"]
)
```

```python
display(
    processed[
        [
            "Product_Name",
            "Unit_Price",
            "Quantity",
            "Total_Amount"
        ]
    ].head(10)
)
```

```python
# Create an order-value category based on Total_Amount
# apply() is used to transform each Total_Amount value
processed["Order_Value_Category"] = processed["Total_Amount"].apply(
    lambda amount:
        "High" if amount >= 5000
        else "Medium" if amount >= 2000
        else "Low" if amount >=1000
    else "Very low"
)
```

```python
# Display the total amount and its corresponding category
display(
    processed[
        [
            "Total_Amount",
            "Order_Value_Category"
        ]
    ].head(10)
)
```

```python
# Right now our DataFrame contains all the information, but we should arrange the columns in a meaningful order.
```

```python
# Define the final column order for the processed dataset
final_columns = [
    "Order_ID",
    "Order_Date",
    "Order_Year",
    "Order_Month",
    "Order_Month_Name",
    "Order_Day",
    "Order_Day_Name",

    "Customer_ID",
    "Customer_Name",
    "City",
    "Region",
    "Membership_Type",

    "Product_ID",
    "Product_Name",
    "Category",
    "Brand",
    "Unit_Price",

    "Quantity",
    "Total_Amount",
    "Order_Value_Category",

    "Payment_Method",
    "Order_Status"
]

# Select the columns in the required order
processed = processed[final_columns]
```

```python
# Sort the processed dataset by Order_Date
# reset_index() creates a clean index after sorting
processed=processed.sort_values(
    "Order_Date"
).reset_index(drop=True)
```
